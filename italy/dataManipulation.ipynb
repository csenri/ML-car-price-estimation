NOW WE NEED TO REMOVE THE EURO SYMBOL FROM THE PRICE AND REMOVE THE DOTS

```python
# need to remove the rows without a price
# 1. keep only rows where 'price' is not NaN
df = df[df["price"].notna()]

# 2. Remove the euro sign and any commas or dots
df["price"] = df["price"].replace({'€': '', r'[.,]': ''}, regex=True)

# 3. remove whute spaces
df["price"] = df["price"].str.strip()

# 4. remove empty strings
df = df[df["price"] != ""]

# 5. convert to integer
df["price"] = df["price"].astype(int) 
df
```

COONVER THE YEAR INTO A SINGLE INTEGER

```python
def convert_year(value):
    # Convert to string and strip spaces
    value = str(value).strip()

    # If it contains any letters, discard it
    if any(char.isalpha() for char in value):
        return None

    # If it's a format like "MM/YYYY"
    if "/" in value:
        parts = value.split("/")
        if len(parts) == 2 and parts[0].isdigit() and parts[1].isdigit():
            month, year = parts
            month = month.zfill(2)
            return int(year + month)
        else:
            return None

    # If it's a pure year like "2025"
    if value.isdigit():
        return int(value + "01")

    return None

# Apply the conversion
df["year"] = df["year"].apply(convert_year)

# Drop rows where the conversion failed (None or NaN)
df = df.dropna(subset=["year"])
df
```
